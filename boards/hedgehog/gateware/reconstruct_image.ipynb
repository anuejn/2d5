```python
# we are reconstructing data from channel 2 of the adc

import numpy as np
data = np.fromfile("/Users/anuejn/tmp/frame3.data", dtype=np.uint8)
```

```python
import matplotlib.pyplot as plt
```

```python
a = data.reshape((-1, 8))[:, 2]
a = a[:len(a)//4*4]
b = data.reshape((-1, 8))[:, 3]
b = b[:len(b)//4*4]
merged = np.empty((len(a) + len(b),))
merged[0::2] = a
merged[1::2] = b

cycles = merged.reshape((-1, 2))
maxes = np.max(cycles, axis=1)
mins = np.min(cycles, axis=1)

plt.plot(mins)
```
